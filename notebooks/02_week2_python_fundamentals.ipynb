## Part 2: Lists

A **list** is an ordered collection of values. Lists use square brackets `[ ]` and commas between items.

```python
my_list = ["apple", "banana", "cherry"]
```

You can access individual items by their **index** (position number). Python starts counting at **0**, not 1.

```python
my_list[0]   # "apple" — first item
my_list[1]   # "banana" — second item
my_list[-1]  # "cherry" — last item
```

````python
# Memphis zip codes
memphis_zips = [38103, 38104, 38105, 38106, 38107, 38108]

print('All zip codes:', memphis_zips)
print('First zip code:', memphis_zips[0])
print('Last zip code:', memphis_zips[-1])
print('Number of zip codes:', len(memphis_zips))
````

## Part 3: Loops

A **loop** lets you repeat an action for each item in a list. The `for` loop syntax is:

```python
for item in my_list:
    # indented code runs once per item
    print(item)
```

**Indentation matters in Python.** Code inside a loop must be indented (4 spaces or 1 Tab). Python uses indentation to know what belongs inside the loop.

````python
# Loop through Memphis neighborhoods and print each one
memphis_neighborhoods = ['Downtown', 'Midtown', 'Cooper-Young', 'Binghampton', 'Orange Mound']

for neighborhood in memphis_neighborhoods:
    print('Neighborhood:', neighborhood)
````

## Part 4: Dictionaries

A **dictionary** stores key-value pairs — like a real dictionary where you look up a word (key) to find its definition (value).

```python
my_dict = {"key1": value1, "key2": value2}
my_dict["key1"]   # returns value1
```

Dictionaries are very useful for storing structured data, like properties of a place.

````python
# A dictionary describing one Memphis location
shelby_farms = {
    'name': 'Shelby Farms Park',
    'type': 'park',
    'area_acres': 4500,
    'zip_code': 38134,
    'has_lake': True
}

print('Park name:', shelby_farms['name'])
print('Area (acres):', shelby_farms['area_acres'])
print('Has a lake?', shelby_farms['has_lake'])
````

## Part 5: Functions

A **function** is a reusable block of code that takes inputs (called **parameters**) and returns an output.

```python
def function_name(parameter1, parameter2):
    # code goes here
    return result
```

Once defined, you can call a function as many times as you want.

````python
# A function that describes a neighborhood
def describe_place(name, population, area_sq_miles):
    density = population / area_sq_miles
    description = f"{name} has {population:,} people in {area_sq_miles} sq miles (density: {density:.0f} people/sq mi)"
    return description

# Call the function with different inputs
print(describe_place('Midtown Memphis', 18400, 4.2))
print(describe_place('Downtown Memphis', 7200, 1.8))
print(describe_place('Whitehaven', 34500, 12.1))
````

````python
# TODO 5: Write a function that calculates a walkability score
# Formula: walkability = (parks * 10) + (bus_stops * 5) - (highways * 15)

def walkability_score(neighborhood_name, parks, bus_stops, highways):
    score = (parks * 10) + (bus_stops * 5) - (highways * 15)   # fill in the formula
    print(f'{neighborhood_name} walkability score: {score}')
    return score

# Test your function
walkability_score('Midtown', parks=8, bus_stops=22, highways=1)
walkability_score('Whitehaven', parks=3, bus_stops=9, highways=3)
````

## Part 6: Loading Real Data with Pandas

**Pandas** is the most important Python library for working with data tables. You will use it in every module of this course.

Key concepts:
- A **DataFrame** is a table with rows and columns (like a spreadsheet)
- `pd.read_csv("filename.csv")` loads a CSV file into a DataFrame
- `df.head()` shows the first 5 rows
- `df.shape` tells you (number of rows, number of columns)
- `df.describe()` gives summary statistics

````python
import pandas as pd

# Load the Memphis mini dataset
df = pd.read_csv(
    'https://raw.githubusercontent.com/youngclick/geoai-memphis/main/data/memphis_mini_points.csv'
)  # loads directly from the course GitHub repo — works in Colab or on your laptop, no download needed

print('Dataset shape:', df.shape)
print(f'  → {df.shape[0]} rows, {df.shape[1]} columns')
print()
print('First 5 rows:')
df.head()
````

````python
# Explore the dataset
print('Column names:', list(df.columns))
print()
print('Data types:')
print(df.dtypes)
print()
print('Basic statistics:')
df.describe()
````

````python
# Count records by category
print('Records per category:')
print(df.groupby('category').size())
print()

# Calculate mean value per category
print('Mean value per category:')
print(df.groupby('category')['value'].mean().round(2))
````

## Part 7: Simple Visualization

Charts help us communicate data patterns. We will use **matplotlib** throughout this course.

A basic bar chart:
```python
import matplotlib.pyplot as plt
data.plot(kind='bar')
plt.show()
```

````python
import matplotlib.pyplot as plt

# Bar chart: mean value by category
category_means = df.groupby('category')['value'].mean()

plt.figure(figsize=(8, 4))
category_means.plot(kind='bar', color='steelblue', edgecolor='white')
plt.title('Mean Value by Category — Memphis Mini Dataset')
plt.xlabel('Category')
plt.ylabel('Mean Value')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
````

````python
# TODO 7: Create a chart showing TOTAL (sum) value per category
# Use color='coral' instead of steelblue

category_totals = df.groupby('category')['value'].sum()  # hint: use .sum() not .mean()

plt.figure(figsize=(8, 4))
category_totals.plot(kind='bar', color='coral', edgecolor='white')
plt.title('Total Values by Category — Memphis Mini Dataset')   # write a good title
plt.xlabel('Category')
plt.ylabel('Value')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
````
